# Project 3: E-commerce Customer Analysis using Polynomial Regression

**Goal:** Explore customer behavior and predict yearly spending from website time using degree-2 Polynomial Regression. This notebook contains illustrative sample data and needs no separate CSV upload.

```python
# Step 1: Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully!")
```

```python
# Step 2: Create example e-commerce customer data
customers = pd.DataFrame({
    "customer_id": range(101, 121),
    "time_on_website_minutes": [10, 12, 15, 18, 20, 22, 24, 25, 27, 29,
                                30, 32, 34, 35, 37, 39, 40, 42, 45, 48],
    "time_on_app_minutes": [20, 22, 25, 28, 30, 32, 29, 34, 35, 31,
                            38, 36, 40, 42, 39, 45, 44, 47, 48, 50],
    "membership_years": [0.5, 0.8, 1.0, 1.2, 1.5, 1.7, 1.8, 2.0, 2.1, 2.3,
                         2.5, 2.6, 2.8, 3.0, 3.1, 3.3, 3.5, 3.7, 4.0, 4.2],
    "yearly_amount_spent": [250, 270, 290, 310, 330, 350, 340, 370, 390, 400,
                            420, 430, 450, 470, 465, 490, 510, 520, 550, 570]
})

print("First five customers:")
display(customers.head())
print("\nSummary statistics:")
display(customers.describe())
```

```python
# Step 3: Explore the numeric columns
sns.pairplot(customers[[
    "time_on_website_minutes",
    "time_on_app_minutes",
    "membership_years",
    "yearly_amount_spent"
]])
plt.suptitle("E-commerce Customer Data Relationships", y=1.02)
plt.show()
```

```python
# Step 4: Choose website time as input and yearly spending as target
X = customers[["time_on_website_minutes"]]
y = customers["yearly_amount_spent"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print("Training customers:", len(X_train))
print("Testing customers:", len(X_test))
```

```python
# Step 5: Train a degree-2 Polynomial Regression model
model = make_pipeline(
    PolynomialFeatures(degree=2, include_bias=False),
    LinearRegression()
)
model.fit(X_train, y_train)

print("Polynomial Regression model training completed!")
```

```python
# Step 6: Evaluate the predictions
predictions = model.predict(X_test)

print("MAE:", round(mean_absolute_error(y_test, predictions), 2))
print("RMSE:", round(mean_squared_error(y_test, predictions) ** 0.5, 2))
print("R-squared:", round(r2_score(y_test, predictions), 3))

results = pd.DataFrame({
    "Actual spending": y_test.values,
    "Predicted spending": predictions.round(2)
})
display(results)
```

```python
# Step 7: Plot customer spending and the polynomial curve
plt.scatter(X["time_on_website_minutes"], y, label="Actual customers")
website_times = pd.DataFrame({
    "time_on_website_minutes": sorted(X["time_on_website_minutes"].tolist())
})
spending_curve = model.predict(website_times)

plt.plot(website_times["time_on_website_minutes"], spending_curve,
         label="Polynomial curve")
plt.xlabel("Time on website (minutes)")
plt.ylabel("Yearly amount spent (sample units)")
plt.title("Website Time vs Yearly Customer Spending")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Step 8: Predict spending for a sample customer
sample_customer = pd.DataFrame({"time_on_website_minutes": [35.0]})
predicted_spending = model.predict(sample_customer)[0]
print("Predicted yearly amount for 35 website minutes:",
      round(predicted_spending, 2))
```
