# Project 1: Spam–Ham Classification

**Goal:** Classify SMS messages as spam or ham. This notebook is ready to upload to Google Colab. The dataset is downloaded in the first code cell.

```python
# Step 1: Import the libraries we need
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Libraries imported successfully!")
```

```python
# Step 2: Download and load the SMS Spam Collection data
data_url = "https://raw.githubusercontent.com/justmarkham/pycon-2016-tutorial/master/data/sms.tsv"

messages = pd.read_csv(
    data_url,
    sep="\t",
    header=None,
    names=["label", "message"]
)

print("First five messages:")
display(messages.head())
print("Total messages:", len(messages))
print("\nLabel counts:")
print(messages["label"].value_counts())
```

```python
# Step 3: Clean the data and choose input/output columns
messages = messages[["label", "message"]].dropna()
messages["label"] = messages["label"].str.lower().str.strip()
messages = messages[messages["label"].isin(["ham", "spam"])]

X = messages["message"]
y = messages["label"]

print("Rows after cleaning:", len(messages))
```

```python
# Step 4: Turn message text into numeric features
vectorizer = TfidfVectorizer(stop_words="english", lowercase=True)
X_features = vectorizer.fit_transform(X)

# Keep some messages for testing the model
X_train, X_test, y_train, y_test = train_test_split(
    X_features, y, test_size=0.20, random_state=42, stratify=y
)

print("Training messages:", X_train.shape[0])
print("Testing messages:", X_test.shape[0])
```

```python
# Step 5: Train a Naive Bayes classifier
model = MultinomialNB()
model.fit(X_train, y_train)

print("Model training completed!")
```

```python
# Step 6: Check how well the model performs
predictions = model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, predictions), 4))
print("\nClassification report:")
print(classification_report(y_test, predictions))

matrix = confusion_matrix(y_test, predictions, labels=["ham", "spam"])
sns.heatmap(
    matrix, annot=True, fmt="d", cmap="Blues",
    xticklabels=["ham", "spam"], yticklabels=["ham", "spam"]
)
plt.xlabel("Predicted label")
plt.ylabel("Actual label")
plt.title("Spam–Ham Confusion Matrix")
plt.show()
```

```python
# Step 7: Try predicting a new message
sample_message = ["Congratulations! You have won a free prize. Call now."]
sample_features = vectorizer.transform(sample_message)
print("Message:", sample_message[0])
print("Prediction:", model.predict(sample_features)[0])
```
