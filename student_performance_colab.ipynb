# Project 2: Student Performance Prediction using Linear Regression

**Goal:** Predict exam marks from study hours. The small sample dataset is created in this notebook, so no separate CSV upload is needed. Replace it with your instructor's dataset if required.

```python
# Step 1: Import libraries
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully!")
```

```python
# Step 2: Create a small example student dataset
students = pd.DataFrame({
    "study_hours": [1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5,
                    5.0, 5.5, 6.0, 6.5, 7.0, 7.5, 8.0, 8.5],
    "attendance_percent": [60, 65, 68, 70, 72, 75, 78, 80,
                           82, 84, 86, 88, 90, 92, 95, 96],
    "previous_score": [45, 48, 50, 52, 55, 58, 60, 62,
                       65, 68, 70, 73, 76, 80, 84, 87],
    "exam_score": [42, 46, 49, 53, 56, 60, 63, 66,
                   70, 73, 77, 80, 84, 88, 93, 96]
})

print("First five rows:")
display(students.head())
print("\nSummary statistics:")
display(students.describe())
```

```python
# Step 3: Select study hours as input and exam score as target
X = students[["study_hours"]]
y = students["exam_score"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
# Step 4: Train the Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

print("Slope:", round(model.coef_[0], 3))
print("Intercept:", round(model.intercept_, 3))
```

```python
# Step 5: Predict test scores and evaluate the model
predicted_scores = model.predict(X_test)

print("MAE:", round(mean_absolute_error(y_test, predicted_scores), 3))
print("RMSE:", round(mean_squared_error(y_test, predicted_scores) ** 0.5, 3))
print("R-squared:", round(r2_score(y_test, predicted_scores), 3))

results = pd.DataFrame({
    "Actual score": y_test.values,
    "Predicted score": predicted_scores.round(2)
})
display(results)
```

```python
# Step 6: Draw the actual scores and regression line
plt.scatter(students["study_hours"], students["exam_score"], label="Actual scores")
ordered_hours = students[["study_hours"]].sort_values("study_hours")
plt.plot(ordered_hours["study_hours"],
         model.predict(ordered_hours), label="Regression line")
plt.xlabel("Study hours per day")
plt.ylabel("Exam score")
plt.title("Study Hours vs Exam Score")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Step 7: Predict a score for a student who studies 5 hours
new_student = pd.DataFrame({"study_hours": [5.0]})
predicted_score = model.predict(new_student)[0]
print("Predicted score for 5 study hours:", round(predicted_score, 2))
```
